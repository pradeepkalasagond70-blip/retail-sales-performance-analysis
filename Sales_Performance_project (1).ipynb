```python
furniture_central = df[
    (df["Category"] == "Furniture") &
    (df["Region"] == "Central")
]
```

```python
import numpy as np
furniture_central.groupby("Product Name")[["Sales", "Profit", "Discount"]].agg({
    "Sales": "sum",
    "Profit": "sum",
    "Discount":"mean"
}).sort_values(by="Profit" , ascending = True)
```

```python
furniture_central["Profit_Margin"] = (
    furniture_central["Profit"] / furniture_central["Sales"]
)
```

```python
furniture_central.groupby("Product Name")["Profit_Margin"].mean().sort_values(ascending=True)
```

```python
product_summary = furniture_central.groupby("Product Name")[["Sales", "Profit", "Discount","Profit_Margin"]].agg({
    "Sales": "sum",
    "Profit": "sum",
    "Discount":"mean",
    "Profit_Margin":"mean"
}).sort_values(by="Profit" , ascending = True)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
X = product_summary[["Discount"]]
y = product_summary["Profit_Margin"]
print(X.shape)
print(y.shape)
```

```python
model.fit(X, y)
model.coef_
```

```python
y_pred = model.predict(X)
```

```python
plt.scatter(X, y)
plt.plot(X, y_pred)
plt.xlabel("Discount")
plt.ylabel("Profit_Margin")
plt.show()
```

```python
model.intercept_
model.score(X, y)
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y, y_pred)
mae
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model_test = LinearRegression()
model_test.fit(X_train, y_train)
```

```python
y_test_pred = model_test.predict(X_test)
y_test_pred = model_test.predict(X_test)
```

```python
r2_test = model_test.score(X_test, y_test)
r2_test
```

```python
mae_test = mean_absolute_error(y_test, y_test_pred)
mae_test
```
